# Initialization

```python
from pypdf import PdfReader
from pathlib import Path
import pandas as pd
import os
import warnings
warnings.simplefilter('ignore')
_p=Path(os.getcwd())
p=_p.parent
```

# For Nightingale2002

```python
hw=[]
ids=[]
for i in os.listdir(p.parent/'data/Nightingale2002/raw/pdf-reports'):
  reader = PdfReader(p.parent/'data/Nightingale2002/raw/pdf-reports'/i)
  page=reader.pages[1]
  line=page.extract_text()
  ids+=[i[2:6]]
  hw+=[s for s in line.splitlines() if s.startswith("Weight:")]

df2002=pd.DataFrame({'test_id':ids,'height_weight':hw})
df2002['weight'] = df2002['height_weight'].str.extract(r'Weight:\s*(\d+|N/A)')
df2002['height'] = df2002['height_weight'].str.extract(r'Height:\s*(\d+|N/A)')
df2002['weight'] = pd.to_numeric(df2002['weight'], errors='coerce')
df2002['height'] = pd.to_numeric(df2002['height'], errors='coerce')
df2002.drop(columns=['height_weight'], inplace=True)
df2002
```

```python
df2002.dropna()
```
