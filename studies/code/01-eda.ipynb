## Initialization

```python
import os
from pathlib import Path
import xarray as xr
import holoviews as hv
from holoviews import opts
import matplotlib.pyplot as plt
import pandas as pd
from great_tables import GT, md, html, loc, style
import helper
import warnings
plt.style.use('default')
plt.rcParams['font.family']='Times New Roman'
plt.rcParams['font.size']=12
warnings.simplefilter("ignore")
hv.extension('bokeh')
_p=Path(os.getcwd())
p=_p.parent
```

```python
helper.extract_lower_cervical()
ds=xr.open_dataset(p/'processed-data/master.nc')
ds.close()
df=pd.read_csv(p/'processed-data/master.csv')

wnkd=pd.read_csv(p.parent/'data/Winkelstein2000/clean-data/a01-specimen-info.csv')
wnkf=pd.read_csv(p.parent/'data/Winkelstein2000/clean-data/a09-flexion-moment.csv')
wnke=pd.read_csv(p.parent/'data/Winkelstein2000/clean-data/a08-extension-moment.csv')
wnkd['Extension']=wnke.A2
wnkd['Flexion']=wnkf.A2
wnkd=wnkd.dropna().reset_index(drop=True)
wnkd['BMI']=wnkd['Weight (kg)']/((wnkd['Height (cm)']/100)**2)
wnkd34=wnkd[wnkd['Cervical Level']=='C3-C4'].reset_index(drop=True)
wnkd56=wnkd[wnkd['Cervical Level']=='C5-C6'].reset_index(drop=True)
```

# Plots and Tables for Nightingale2002+Nightingale2007

## Count Table

```python
c=pd.DataFrame()
c['l']=['C3C4', 'C4C5', 'C5C6', 'C6C7']
c['mf']=df[(df['sex']=='Male')&(df['loading']=='Flexion')].value_counts('level',sort=False).values
c['me']=df[(df['sex']=='Male')&(df['loading']=='Extension')].value_counts('level',sort=False).values
c['ff']=df[(df['sex']=='Female')&(df['loading']=='Flexion')].value_counts('level',sort=False).values
c['fe']=df[(df['sex']=='Female')&(df['loading']=='Extension')].value_counts('level',sort=False).values
(
    GT(c,rowname_col='l')
    .tab_spanner(label=md("*Male*"), columns=["mf", "me"])
    .tab_spanner(label=md("*Female*"), columns=["ff", "fe"])
    .tab_header(
        title=html("Data Count - Nightingale ")
    )
    .cols_label(mf="Flexion", me="Extension",ff="Flexion", fe="Extension")
    .tab_stubhead('Cervical Level')
    .fmt_integer(columns=["mf", "me"], use_seps=False)
    .tab_source_note(source_note="Only Lower Cervical Flexibility Tests' Data")
    .tab_style(
        style=style.fill(color="lightcyan"),
        locations=loc.body(columns=["mf","me"]),
    )
    .tab_style(
        style=style.fill(color="mistyrose"),
        locations=loc.body(columns=["ff","fe"]),
    )
)
```
