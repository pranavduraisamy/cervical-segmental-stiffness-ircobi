# Interpolating data

```python
import os
from pathlib import Path
import pandas as pd
import numpy as np
import xarray as xr
import holoviews as hv
hv.extension('bokeh')
_p=Path(os.getcwd())
p=_p.parent
```

```python
cat=pd.read_csv(p/'clean-data/data-catalog.csv')
cat=cat[(cat['ascii_id']==1) & (cat['test']=='flexibility')].copy()
cat.drop('injury', axis=1, inplace=True)
cat=cat.reset_index(drop=True)

temp = []
flex=cat[(cat['loading']=='extension')|(cat['loading']=='flexion')].copy().reset_index(drop=True)
b=list(cat[cat['loading']=='both'].copy()['test_id'])
for index, row in cat.iterrows():
    if row['loading'] == 'both':
        rf=row.copy()
        rf['loading'] = 'flexion'
        temp.append(rf)
        re = row.copy()
        re['loading'] = 'extension'
        temp.append(re)
flex=pd.concat([flex,pd.DataFrame(temp)],ignore_index=True)

r=[]
xn_col=["%.1f" % round(i, 1) for i in np.linspace(0,3.5,36)]
for i in range(0,len(flex)):
    tid=flex['test_id'][i]
    dfx=pd.read_csv(p/str('raw/extracted/b0'+str(tid)+'/b0'+str(tid)+'.001'),sep='\t',header=None)
    xn=np.linspace(0,3.5,36)
    temp=[tid,flex['loading'][i]]
    if(flex['loading'][i]=='flexion' and tid in b):     #Flexion case - both
        dfx=dfx[dfx[0]>=0].reset_index(drop=True)
        y_intp = np.interp(xn, dfx[1], dfx[0])

    elif(flex['loading'][i]=='extension' and tid in b):     #Extension case - both
        dfx=dfx[dfx[0]<=0]
        dfx= dfx.iloc[::-1].reset_index(drop=True)
        dfx[0]=dfx[0].apply(lambda x0: abs(x0))
        dfx[1]=dfx[1].apply(lambda x0: abs(x0))
        y_intp = np.interp(xn, dfx[1], dfx[0])
    y_intp[(xn>list(dfx[1])[-1])]=np.nan
    temp+=list(y_intp)
    r.append(temp)

xn=['test_id','loading']+xn_col
mstr=pd.merge(flex,pd.DataFrame(r,columns=xn),on=['test_id','loading'],how='outer')
mstr=mstr[~mstr['test_id'].isin([4388,4390])].reset_index(drop=True)
rot=mstr.copy()
mstr.insert(loc=0,column='new_id',value=mstr['test_id'].astype(str)+mstr['loading'].str[0])
mstr=mstr[~mstr['new_id'].isin(['4427f'])].reset_index(drop=True)
rot=mstr.copy()
mstr.to_csv(p/'clean-data/master.csv',index=None)
xn=xn_col
rot=rot[xn].copy()
```

```python
ds = xr.Dataset(
    {
        "rotation": (["test_id","moment"],rot.to_numpy()),
        "ref_id": (["test_id"],mstr['ref_id']),
        "sample_id": (["test_id"],mstr['sample_id']),
        "sex": (["test_id"],mstr['sex']),
        "age": (["test_id"],mstr['age']),
        "level": (["test_id"],mstr['level']),
        "loading": (["test_id"],mstr['loading']),
        "cod": (["test_id"],mstr['cod'])
    },
    coords={
        "test_id":mstr['new_id'],
        "moment": xn
    }
)
ds.to_netcdf(p/'clean-data/dataset.nc')
```

```python
def curv(temp):                    
    return hv.Curve((ds.rotation.sel(test_id=temp).values,ds.moment.values)).opts(xlabel='Rotation',ylabel='Moment').redim(y=hv.Dimension('Moment',range=(0,3.5)))
curve_dict={temp:curv(temp) for temp in ds.test_id.values}
NdLayout1=hv.NdLayout(curve_dict, kdims='test_id')
NdLayout1
```
