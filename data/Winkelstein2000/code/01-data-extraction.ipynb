```python
from tabula import read_pdf
import pandas as pd
from pathlib import Path
import os
import numpy as np
import warnings
warnings.simplefilter('ignore')
_p=Path(os.getcwd())
p=_p.parent
```

```python
def helper(pn,flag,x):
    col=['Specimen','F1','F2','F3','M1','M2','M3','X1','X2','X3','A1','A2','A3']
    if flag==1:                             #top
        area=[56.88,21.6,380.16,576.72]
    elif flag==2:                           #bottom
        area=[428,21.6,737,576.72]
    elif flag==3:                           #donor info
        area=[107,21.6,264,355]
    elif flag==4:                           #pg 17 bottom
        area=[358,21.6,737,668]
    pdf=read_pdf(p/'raw/Winkelstein2000.pdf',pages=pn,stream=True,area=area)
    df=pd.DataFrame(pdf[0])
    if flag!=3:
        df.columns=col
        df=df.drop(0)
        df.reset_index(drop=True)
    if flag==3:
        df[['Age', 'Sex']]=df['Age/Sex'].str.split('/', expand=True)
        df.drop(columns=['Age/Sex'], inplace=True)
    df.replace('N/A',np.nan,inplace=True)
    df.to_csv(p/str('clean-data/'+x+'.csv'),index=False)
```

```python
helper(17,3,'a01-specimen-info')
helper(17,4,'a02-posteroanterior-shear')
helper(18,1,'a03-anteroposterior-shear')
helper(18,2,'a04-lateral-shear')
helper(19,1,'a05-compression')
helper(19,2,'a06-tension')
helper(20,1,'a07-lateral-bending')
helper(20,2,'a08-extension-moment')
helper(21,1,'a09-flexion-moment')
helper(21,2,'a10-axial-torsion')
```
