## Initialization

```python
import requests
import os
from pathlib import Path
import zipfile
import warnings
warnings.simplefilter("ignore")
_p=Path(os.getcwd())
p=_p.parent
```

```python
def urls_lst(x,frmt='.zip'):
    urls_x=[]
    if(frmt=='.zip'):
        for j in range(0,len(x),2):
            for i in range(x[j],x[j+1]):
                tempie="https://nrd-static.nhtsa.dot.gov/compress/ev/biodb/b00000/b0".strip()+str(i)[0:2]+"00/b0".strip()+str(i)+'.zip'
                urls_x.append(tempie)
    elif(frmt=='.pdf'):
        for j in range(0,len(x),2):
            for i in range(x[j],x[j+1]):
                tempie="https://nrd-static.nhtsa.dot.gov/reports/biodb/b00000/b0".strip()+str(i)[0:2]+"00/b0".strip()+str(i)+'R001.pdf'
                urls_x.append(tempie)
    return urls_x

def downloader(urls,l):
    if not os.path.exists(l):
        os.makedirs(l)
    for url in urls:
        resp = requests.get(url)
        if resp.status_code==200:
            fn = url.split('/')[-1]
            with open(os.path.join(l, fn), 'wb') as f:
                f.write(resp.content)
            print(f"Downloaded: {fn}")
        else:
            print(f"Failed to download: {url}")
```

## Downloading from NHTSA

```python
m=[7356,7365,7398,7509,6355,6363]
urls_m=urls_lst(m)
urls_p=urls_lst(m,'.pdf')
downloader(urls_m,p/'raw/ascii/')
downloader(urls_p,p/'raw/pdf-reports/')
```

## Unzipping ASCII files

```python
for i in os.listdir(p/'raw/ascii'):
    with zipfile.ZipFile(p/str('raw/ascii/'+i)) as fle:
        fle.extractall(p/str('raw/extracted/'+i[0:6]))
```
